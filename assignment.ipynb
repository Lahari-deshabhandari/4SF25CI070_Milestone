```python
pf.info()
```

```python
pf.drop(['PaymentMethod'], axis=1)
```

```python
pf.loc[0:5,['PaymentMethod']]  #axis is column here
```

```python
pf.drop(['PaymentMethod'], axis=1,inplace=True)
```

```python
pf['ShipCountry']=pf['ShipCountry'].replace('China','Zhongguo')
pf
```

```python
pf['rupees']=pf['TotalAmount'].astype(str).str.split('.').str[0]
pf['paise']=pf['TotalAmount'].astype(str).str.split('.').str[1]
pf
```

```python
pf['ShipCountry'].unique()
```

```python
pf.groupby('ShipCountry')['ShippingFee'].sum().sort_values(ascending=False)
```

```python
pf.drop(columns=['Unnamed: 6','Unnamed: 7','Unnamed: 8'],errors='ignore',inplace=True)
```

```python
pf['TotalAmount'].mean()
```

```python
pf['TotalAmount']=pf['TotalAmount'].fillna(pf['TotalAmount'].mean())
```

```python
pf.drop(['TotalAmount'],axis=1,inplace=True)
```

```python
pf['ShippingFee'].mean()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8,5))
plt.scatter(pf['ShippingFee'], pf['TotalAmount'])

plt.xlabel('Shipping Fee')
plt.ylabel('Total Amount')
plt.title('Shipping Fee vs Total Amount')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(pf['TotalAmount'], bins=20, edgecolor='black')

plt.title('Distribution of Total Amount')
plt.xlabel('Total Amount')
plt.ylabel('Frequency')

plt.show()
```

```python
pd.crosstab(pf['OrderStatus'], pf['PaymentMethod']).plot(kind='bar', stacked=True, figsize=(8,5))
plt.title('Order Status by Payment Method')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=pf, x='ShippingFee', y='rupees', hue='OrderStatus', palette='viridis')
plt.title('Shipping Fee vs Order Amount')
plt.xlabel('Shipping Fee')
plt.ylabel('Order Amount (rupees)')
plt.show()
```

```python
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(6,6))
plt.title('Gender Share')
plt.ylabel('')
plt.show()
```

```python
df['Gender'].value_counts().plot(kind='bar', color='steelblue', figsize=(8,4))
plt.title('Customers by Gender')
plt.xlabel('Gender'); plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
sns.catplot(data=pf, x='OrderStatus', y='ShippingFee', kind='box')
plt.show()
```

```python
df['Country'].value_counts().head(10).plot(kind='bar', figsize=(9,5))

plt.title('Top 10 Countries by Customers')
plt.xlabel('Country')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.show()
```

```python
df['Age'] = 2026 - df['DateOfBirth'].str.split('/').str[2].astype(int)

sns.kdeplot(df['Age'].dropna(), fill=True)

plt.title('Distribution of Customer Age')
plt.xlabel('Age')
plt.ylabel('Density')

plt.show()
```
